# Hands-on Activity 9.1 Tensorflow Exercises

### Import

```python
import tensorflow as tf
print(tf.__version__)
```

### 1. Create a vector, scalar, matrix and tensor with values of your choosing using tf.constant().

```python
scalar = tf.constant(7)
vector = tf.constant([6,7])
matrix = tf.constant([[10,7],[6,7]])
tensor = tf.constant([[[1,2,3],[4,5,6]],[[5,6,7],[4,3,1]]])
```

### 2. Find the shape, rank and size of the tensors you created in 1.

```python
print("Shapes")
print(tf.shape(scalar))
print(tf.shape(vector))
print(tf.shape(matrix))
print(tf.shape(tensor))
```

```python
print("Ranks")
print(tf.rank(scalar))
print(tf.rank(vector))
print(tf.rank(matrix))
print(tf.rank(tensor))
```

```python
print("Size")
print(tf.size(scalar))
print(tf.size(vector))
print(tf.size(matrix))
print(tf.size(tensor))
```

### 3. Create two tensors containing random values between 0 and 1 with shape [5, 300].

```python
X = tf.random.uniform(shape=[5, 300], minval=0, maxval=1, dtype=tf.float32)
Y = tf.random.uniform(shape=[5, 300], minval=0, maxval=1, dtype=tf.float32)
```

```python
print(tf.shape(X))
print(tf.shape(Y))
```

### 4. Multiply the two tensors you created in 3 using matrix multiplication.

```python
matrix_result = tf.math.multiply(X, Y)
matrix_result.shape
```

### 5. Multiply the two tensors you created in 3 using dot product.

```python
dot_result = tf.tensordot(X, tf.transpose(Y), axes=1)
dot_result.shape
```

### 6. Create a tensor with random values between 0 and 1 with shape [224, 224, 3].

```python
A = tf.random.uniform(shape=[224, 224, 3], minval=0, maxval=1)
```

### 7. Find the min and max values of the tensor you created in 6.

```python
A_max = tf.reduce_max(A).numpy()
A_min = tf.reduce_min(A).numpy()

print(f"Max Value in A: {A_max}")
print(f"Min Value in A: {A_min}")
```

### 8. Created a tensor with random values of shape [1, 224, 224, 3] then squeeze it to change the shape to [224, 224, 3].

```python
B1 = tf.random.uniform(shape=[1, 224, 224, 3])
B2 = tf.reshape(B1, [224, 224, 3])
print(B1.shape)
print(B2.shape)
```

### 9. Create a tensor with shape [10] using your own choice of values, then find the index which has the maximum value.

```python
c = [7, 6, 24, 5, 11, 14, 6, 100, 101, 102]
C = tf.constant(c)
C.shape
```

```python
C_max_index = tf.argmax(C)
print(f"Max value index: {C_max_index.numpy()}")
```

### 10. One-hot encode the tensor you created in 9.

```python
depth = 3
C_one_hot = tf.one_hot(C, depth)
C_one_hot
```
